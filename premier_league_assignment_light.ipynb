# Premier League Data Visualization Practice

Dataset: `premier_league_practice_dataset.csv` — 200 records (20 teams x 10 seasons: 2015-16 to 2024-25).

Tools: Pandas, Matplotlib, Seaborn. Each chart has a title, axis labels, and a short observation.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

df = pd.read_csv('premier_league_practice_dataset.csv')
season_order = ['2015-16','2016-17','2017-18','2018-19','2019-20','2020-21','2021-22','2022-23','2023-24','2024-25']
df['Season'] = pd.Categorical(df['Season'], categories=season_order, ordered=True)
print(df.shape)
df.head()
```

```python
df.info()
df.describe().round(2)
```

```python
print(df['Team'].nunique())
print(sorted(df['Team'].unique()))
print(sorted(df['Season'].unique()))
```

## Part 1: Line Charts

### Q1. Points earned by Arsenal across all seasons

```python
ars = df[df['Team']=='Arsenal'].sort_values('Season')
plt.figure()
plt.plot(ars['Season'].astype(str), ars['Points'], marker='o')
plt.title('Arsenal Points Across Seasons')
plt.xlabel('Season')
plt.ylabel('Points')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print(ars[['Season','Points']].to_string(index=False))
```

**Observation:** Arsenal is highly inconsistent — peaks of 90 pts (2016-17) and 86 pts (2019-20) vs lows of 49–51 pts (2020-21, 2017-18, 2023-24). Finished 2024-25 with 71 pts.

### Q2. Compare points of Arsenal, Liverpool and Manchester City

```python
sub = df[df['Team'].isin(['Arsenal','Liverpool','Manchester City'])].sort_values(['Team','Season'])
plt.figure()
for team, g in sub.groupby('Team'):
    g = g.sort_values('Season')
    plt.plot(g['Season'].astype(str), g['Points'], marker='o', label=team)
plt.title('Arsenal vs Liverpool vs Manchester City — Points by Season')
plt.xlabel('Season')
plt.ylabel('Points')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Observation:** No team dominates every season. Arsenal peaked highest (90 in 2016-17), Man City peaked late (86 in 2022-23, 78 in 2024-25), Liverpool is the most stable (53–77 range, 72 in 2024-25).

### Q3. Goals scored (Goals_For) by Tottenham

```python
tot = df[df['Team']=='Tottenham'].sort_values('Season')
plt.figure()
plt.plot(tot['Season'].astype(str), tot['Goals_For'], marker='o')
plt.title('Tottenham Goals Scored Across Seasons')
plt.xlabel('Season')
plt.ylabel('Goals For')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print(tot[['Season','Goals_For']].to_string(index=False))
```

**Observation:** Tottenham scoring fluctuates 39–61 goals; best spells 2021-22 (61) and 2023-24 (60), weakest 2020-21 (39). 2024-25 scored 47.

### Q4. Possession % of Chelsea, Liverpool, Man City

```python
sub = df[df['Team'].isin(['Chelsea','Liverpool','Manchester City'])].sort_values(['Team','Season'])
plt.figure()
for team, g in sub.groupby('Team'):
    g = g.sort_values('Season')
    plt.plot(g['Season'].astype(str), g['Possession_Percent'], marker='o', label=team)
plt.title('Possession % — Chelsea vs Liverpool vs Man City')
plt.xlabel('Season')
plt.ylabel('Possession %')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Observation:** All three usually hold 44–64%. Man City hit 62.2% (2023-24). Chelsea is generally lowest of the three (e.g. 43.5% in 2022-23). Possession does not strictly track points.

### Q5. Average points for all teams in each season

```python
avg_pts = df.groupby('Season', observed=True)['Points'].mean().reindex(season_order)
plt.figure()
plt.plot(avg_pts.index.astype(str), avg_pts.values, marker='o')
plt.title('Average Points per Season (All Teams)')
plt.xlabel('Season')
plt.ylabel('Average Points')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print(avg_pts.round(2))
```

**Observation:** League average rose from 45.85 (2015-16, lowest) to a peak of 58.8 (2022-23), then stabilized ~55–56.4 in 2023-24/2024-25 — overall competitiveness/points inflation increased.

### Q6. Clean sheets by Chelsea

```python
che = df[df['Team']=='Chelsea'].sort_values('Season')
plt.figure()
plt.plot(che['Season'].astype(str), che['Clean_Sheets'], marker='o')
plt.title('Chelsea Clean Sheets Across Seasons')
plt.xlabel('Season')
plt.ylabel('Clean Sheets')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print(che[['Season','Clean_Sheets']].to_string(index=False))
```

**Observation:** Chelsea defense is volatile: best 17 clean sheets (2018-19), then 16 (2024-25); worst 7 (2019-20, 2023-24). Two strong defensive seasons bookend the period.

### Q7. Wins for Arsenal vs Manchester United

```python
sub = df[df['Team'].isin(['Arsenal','Manchester United'])].sort_values(['Team','Season'])
plt.figure()
for team, g in sub.groupby('Team'):
    g = g.sort_values('Season')
    plt.plot(g['Season'].astype(str), g['Wins'], marker='o', label=team)
plt.title('Wins — Arsenal vs Manchester United')
plt.xlabel('Season')
plt.ylabel('Wins')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Observation:** Both swing widely. Arsenal highs 27 wins (2016-17, 2019-20); United highs 24 (2024-25) and 23 (2021-22). United was stronger most recently (24 vs 20 wins in 2024-25).

## Part 2: Bar Charts

### Q8. Top 10 teams by points in 2024-25

```python
d = df[df['Season']=='2024-25'].sort_values('Points', ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(data=d, x='Points', y='Team', hue='Team', legend=False, palette='viridis')
plt.title('Top 10 Teams by Points — 2024-25')
plt.xlabel('Points')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
print(d[['Team','Points']].to_string(index=False))
```

**Observation:** Man City (78) edges Man United (77), then Liverpool and Leicester (72 each). Gap from 1st to 10th (Everton 56) is 22 pts — top-heavy season.

### Q9. 10 teams with lowest points in 2023-24

**Observation:** Fulham (24) is a clear outlier at the bottom, followed by Bournemouth (32). The 10th-lowest (Crystal Palace 57) would be mid-table — big gap between relegated and rest.

### Q10. Goals scored by top 8 teams (season 2024-25)

```python
season = '2024-25'
d = df[df['Season']==season].sort_values('Points', ascending=False).head(8)
plt.figure(figsize=(10,5))
sns.barplot(data=d, x='Team', y='Goals_For', hue='Team', legend=False, palette='rocket')
plt.title(f'Goals Scored by Top 8 Teams — {season}')
plt.xlabel('Team')
plt.ylabel('Goals For')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
print(d[['Team','Goals_For']].to_string(index=False))
```

**Observation:** Among the top 8, Liverpool (72) and Brighton (64) outscore champions Man City (52) — City won with defense/consistency, not goals. Scoring rank ≠ points rank.

### Q11. Average possession for top 10 teams by points in 2024-25

```python
d = df[df['Season']=='2024-25'].sort_values('Points', ascending=False).head(10).sort_values('Possession_Percent')
plt.figure(figsize=(10,5))
sns.barplot(data=d, x='Possession_Percent', y='Team', hue='Team', legend=False, palette='Blues_r')
plt.title('Possession % of Top 10 Teams by Points — 2024-25')
plt.xlabel('Possession %')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
print(d[['Team','Possession_Percent']].to_string(index=False))
```

**Observation:** Fulham finished 5th (71 pts) with the lowest possession (37.2%) of the top 10, while Newcastle (60.8%) finished 8th — possession alone does not explain ranking.

### Q12. Wins, Draws, Losses for one team in one season (Arsenal, 2024-25)

```python
row = df[(df['Team']=='Arsenal') & (df['Season']=='2024-25')].iloc[0]
vals = pd.Series({'Wins': row['Wins'], 'Draws': row['Draws'], 'Losses': row['Losses']})
plt.figure()
sns.barplot(x=vals.index, y=vals.values, hue=vals.index, legend=False, palette='Set2')
plt.title('Arsenal 2024-25 — Wins, Draws, Losses')
plt.ylabel('Matches')
plt.tight_layout()
plt.show()
print(vals)
```

**Observation:** Arsenal 2024-25: 20 wins, 11 draws, 7 losses in 38 games — draws (11) cost the title despite few defeats.

### Q13. Top 10 clean sheets in 2022-23

```python
d = df[df['Season']=='2022-23'].sort_values('Clean_Sheets', ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(data=d, x='Clean_Sheets', y='Team', hue='Team', legend=False, palette='Greens_r')
plt.title('Top 10 Clean Sheets — 2022-23')
plt.xlabel('Clean Sheets')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
print(d[['Team','Clean_Sheets']].to_string(index=False))
```

### Q14. Average yellow cards per season

```python
d = df.groupby('Season', observed=True)['Yellow_Cards'].mean().reindex(season_order)
plt.figure()
sns.barplot(x=d.index.astype(str), y=d.values, hue=d.index.astype(str), legend=False, palette='YlOrBr')
plt.title('Average Yellow Cards per Season')
plt.xlabel('Season')
plt.ylabel('Avg Yellow Cards')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print(d.round(1))
```

**Observation:** Discipline is stable: 62–69 cards average. Peak 68.6 (2017-18), lowest 62.2 (2016-17). No strong trend over time.

### Q15. Top 10 passing accuracy in 2024-25

```python
d = df[df['Season']=='2024-25'].sort_values('Pass_Accuracy_Percent', ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(data=d, x='Pass_Accuracy_Percent', y='Team', hue='Team', legend=False, palette='coolwarm')
plt.title('Top 10 Passing Accuracy — 2024-25')
plt.xlabel('Pass Accuracy %')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
print(d[['Team','Pass_Accuracy_Percent']].to_string(index=False))
```

### Q16. Goals For vs Goals Against for five teams (top 5, 2024-25)

```python
d = df[df['Season']=='2024-25'].sort_values('Points', ascending=False).head(5)
m = d[['Team','Goals_For','Goals_Against']].melt(id_vars='Team', var_name='Type', value_name='Goals')
plt.figure(figsize=(10,5))
sns.barplot(data=m, x='Team', y='Goals', hue='Type')
plt.title('Goals For vs Goals Against — Top 5 Teams 2024-25')
plt.xlabel('Team')
plt.ylabel('Goals')
plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()
print(d[['Team','Goals_For','Goals_Against']].to_string(index=False))
```

**Observation:** Fulham (56–22) has the best defense of the top 5; both Manchester clubs conceded more than they scored yet finished 1st–2nd — points came from close wins. Liverpool (72–53) is the most attacking.

## Part 3: Scatter Plots

### Q17. Shots per Match vs Goals For

```python
plt.figure()
sns.scatterplot(data=df, x='Shots_Per_Match', y='Goals_For')
plt.title('Shots per Match vs Goals Scored')
plt.xlabel('Shots per Match')
plt.ylabel('Goals For')
plt.tight_layout()
plt.show()
print('corr =', round(df[['Shots_Per_Match','Goals_For']].corr().iloc[0,1], 3))
```
